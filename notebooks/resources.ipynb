### 📚 Extra Resources!

A collection of resources for learning about linear algebra, Kalman Filters, Extended Kalman Filters

This is as a hand pick list of resources that are helpful! 

### Linear Algebra: 

- [3blue1brown's Essence of Linear Algebra](https://youtube.com/playlist?list=PLZHQObOWTQDPD3MizzM2xVFitgF8hE_ab&si=7Zo2aclopxMwrkn8) -> great for visualization of many linear algebra topics!

- [Gilbert Strang's MiT Open Coursework](https://youtube.com/playlist?list=PLE7DDD91010BC51F8&si=q6u6RjmYRql-3qu9) -> the holy grail for all linear algebra students

### Kalman Filters: 

- [Explained Simply](https://thekalmanfilter.com/kalman-filter-explained-simply/) -> great for understanding conceptually how Kalman filter work 

- [Kalmanfilter.net](https://kalmanfilter.net/) -> holy grail of Kalman filtering! 

- [A more formal approach](https://www.unitedthc.com/DSP/Kalman1960.pdf) -> the original Kalman Filter paper

### Extended Kalman Filters: 

- [Really all you need](https://thekalmanfilter.com/extended-kalman-filter-python-example/) -> amazing! 

- [A more formal approach](https://ntrs.nasa.gov/api/citations/19860003843/downloads/19860003843.pdf) -> the original EKF paper
